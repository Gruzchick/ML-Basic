### Создайте одномерный массив размера 10, заполненный нулями и пятым элемент равным 1. Трансформируйте в двумерный массив.

### Создайте одномерный массив со значениями от 10 до 49 и разверните его (первый элемент становится последним). Найдите в нем все четные элементы.

```python
arr = np.arange(10, 50)

arr_reversed = arr[::-1]

even_elements = arr_reversed[arr_reversed % 2 == 0]

even_elements
```

### Создайте массив 4x3x2 со случайными значениями. Найти его минимум и максимум.

```python
random_arr = np.random.random((4, 3, 2))

min_val = random_arr.min()
max_val = random_arr.max()

print(f"Min: {min_val}")
print(f"Max: {max_val}")
random_arr
```

### Создайте два двумерных массива размерами 6x4 и 4x3 и произведите их матричное умножение. 

```python
m1 = np.random.random((6, 4))
m2 = np.random.random((4, 3))

result = m1 @ m2

result
```

### Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.

```python
arr = np.random.random((7, 7))

mean_val = arr.mean()
std_val = arr.std()

arr_normalized = (arr - mean_val) / std_val

print(f"Среднее: {mean_val:.4f}")
print(f"Стд. отклонение: {std_val:.4f}")
print("-" * 20)
print(f"Среднее: {arr_normalized.mean():.4f}")
print(f"Стд. отклонение: {arr_normalized.std():.4f}")

arr_normalized
```

```python
(tips['smoker'] == 'Yes').sum()
```

### Узнайте какой средний 'total_bill' в зависимости от 'day'

```python
tips.groupby('day', observed=True)['total_bill'].mean()
```

### Отберите строчки с 'total_bill' больше медианы и узнайте какой средний 'tip' в зависимости от 'sex'

```python
median_bill = tips['total_bill'].median()

high_bill_tips = tips[tips['total_bill'] > median_bill]

high_bill_tips.groupby('sex', observed=True)['tip'].mean()
```

### Преобразуйте признак 'smoker' в бинарный (0-No, 1-Yes)

```python
tips['smoker'] = tips['smoker'].map({'Yes': 1, 'No': 0})

tips.head()
```

# III. Visualization

### Постройте гистограмму распределения признака 'total_bill'

```python
sns.histplot(data=tips, x='total_bill', kde=True)

plt.title('Распределение суммы чека (Total Bill)')
plt.show()
```

### Постройте scatterplot, представляющий взаимосвязь между признаками 'total_bill' и 'tip'

```python
sns.scatterplot(data=tips, x='total_bill', y='tip')
plt.title('Связь чека и чаевых')
plt.show()
```

### Постройте pairplot

```python
sns.pairplot(tips)
plt.show()
```

### Постройте график взаимосвязи между признаками 'total_bill' и 'day'

```python
sns.boxplot(data=tips, x='day', y='total_bill')

plt.title('Распределение суммы чека по дням')
plt.show()
```

### Постройте две гистограммы распределения признака 'tip' в зависимости от категорий 'time'

```python
sns.displot(data=tips, x='tip', col='time', kde=True)

plt.show()
```

### Постройте два графика scatterplot, представляющих взаимосвязь между признаками 'total_bill' и 'tip' один для Male, другой для Female и раскрасьте точки в зависимоти от признака 'smoker'

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    col='sex',
    hue='smoker'
)

plt.show()
```
